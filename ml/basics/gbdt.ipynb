# GBDT

Gradient Boosting — sequential regression trees fit pseudo-residuals (sin(x) reg + binary cls).

## 背景

Gradient Boosting — sequential regression trees fit pseudo-residuals (sin(x) reg + binary cls). 本 notebook 演示 GBDT 的完整实现，模型代码见 `ml/basics/gbdt.py`。


## 数学原理

**初始化：** $F_0(x) = \arg\min_\gamma \sum L(y_i, \gamma)$

**对 m=1..M：**
  1. 负梯度（伪残差）: $\tilde{y}_i = -\left[\frac{\partial L(y_i, F(x_i))}{\partial F(x_i)}\right]_{F=F_{m-1}}$
  2. 拟合回归树: $h_m(x)$ 到 $\tilde{y}_i$
  3. 更新: $F_m(x) = F_{m-1}(x) + \eta \cdot h_m(x)$
