## 数学原理

**中心化：** $\tilde{X} = X - \bar{x}$

**SVD：** $\tilde{X} = U \Sigma V^T$

**投影到前 k 个主成分：** $X_{\text{proj}} = \tilde{X} \cdot V_{:,:k}$
