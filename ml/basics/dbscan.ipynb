## 数学原理

**eps-邻域：**

$$N_\varepsilon(p) = \{ q \in D \mid \mathrm{dist}(p, q) \le \varepsilon \}$$

**三类点：**

- **核心点（core）**：$|N_\varepsilon(p)| \ge \text{min\_samples}$
- **边界点（border）**：不是核心点，但落在某个核心点的 $\varepsilon$-邻域内
- **噪声（noise）**：两者都不是，标为 $-1$

**簇：** 从一个核心点出发、由密度可达（density-reachable）关系连成的最大点集。
